# Read the ice cream sales dataset from CSV file for Polynomial Regression

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
df=pd.read_csv('Ice_cream selling data.csv')
df
```

```python
df.columns
```

```python
# Select the first column as input feature (X)
# 0:1 keeps X in 2D shape, which is required for regression models
x=df.iloc[:,0:1].values

# Select the second column as output/target variable (y)
# y is kept as 1D array as required by scikit-learn
y=df.iloc[:,1].values
```

```python
#Fitting linear regression
lr=LinearRegression()
lr.fit(x,y)
```

```python
#Fitting Polynomial Regression
from sklearn.preprocessing import PolynomialFeatures
poly=PolynomialFeatures(degree=4)
x_poly=poly.fit_transform(x)
```

```python
plt.scatter(x,y,color='red')

plt.plot(x, lr.predict(x), color='blue')
plt.title('Linear Regression')
plt.xlabel('Temperature')
plt.ylabel('Ice Cream sales')

plt.show()
```

```python
X_poly = poly.fit_transform(x)
lr.fit(X_poly, y)
```

```python
plt.scatter(x, y, color='blue')
plt.plot(x, lr.predict(poly.fit_transform(x)),
         color='red')
plt.title('Polynomial Regression')
plt.xlabel('Temperature')
plt.ylabel('Ice Cream sales')

plt.show()
```
